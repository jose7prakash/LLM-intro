```python
two_col = np.array(bank[["age", "Cust_num", "balance", "loan"]])
print(two_col)
print('shape =' + str(two_col.shape))
print('dimensions =' + str(two_col.ndim))
print('data type =' + str(two_col.dtype))
print('size =' + str(two_col.size))
```
